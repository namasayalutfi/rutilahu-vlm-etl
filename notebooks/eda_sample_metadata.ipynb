```python
from pathlib import Path
import json
import pandas as pd
import matplotlib.pyplot as plt

# Path file
input_path = Path(r"C:\Users\Lutfi\Documents\Project\AITF\rutilahu-vlm-etl\metadata\mkn2_metadata_merged.json")

# --- PERBAIKAN DI SINI ---
# Pandas bisa langsung membaca file JSON secara otomatis
try:
    # Coba baca sebagai JSON standar (Array of Objects)
    df = pd.read_json(input_path, encoding="utf-8")
except ValueError:
    # Jika ternyata file kamu memang format JSONL (JSON Lines), pakai lines=True
    df = pd.read_json(input_path, encoding="utf-8", lines=True)
# -------------------------

# Hitung persebaran house_type
# Menggunakan dropna=False agar data yang kosong (NaN) tetap terhitung
house_type_counts = df["house_type"].value_counts(dropna=False)

# Tampilkan jumlah pasti
print("Persebaran house_type:")
print(house_type_counts)
print("\nTotal data:", len(df))

# Chart
plt.figure(figsize=(8, 5))
house_type_counts.plot(kind="bar", color="skyblue", edgecolor="black")
plt.title("Persebaran House Type")
plt.xlabel("House Type")
plt.ylabel("Jumlah")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```
