```python
import json
import pandas as pd

# path file jsonl
jsonl_path = "../metadata/metadata.jsonl"

# path output txt
output_txt = "../metadata/house_id_atap_bambu.txt"

# load jsonl
data = []
with open(jsonl_path, "r", encoding="utf-8") as f:
    for line in f:
        data.append(json.loads(line))

# flatten data
rows = []
for item in data:
    rows.append({
        "house_id": item["house_id"],
        "house_type": item["house_type"],
        "atap": item["actual_label"]["atap"],
        "dinding": item["actual_label"]["dinding"],
        "lantai": item["actual_label"]["lantai"]
    })

df = pd.DataFrame(rows)

print("=" * 80)
print("TOTAL DATA")
print("=" * 80)
print(df["house_type"].value_counts())

# =====================================================
# Filter rumah dengan atap bambu
# =====================================================
rumah_atap_bambu = df[
    df["atap"].fillna("").str.lower() == "bambu"
]

print("\n" + "=" * 80)
print("HOUSE ID DENGAN ATAP BAMBU")
print("=" * 80)

print(f"Total rumah dengan atap bambu: {len(rumah_atap_bambu)}")

# Simpan ke file txt
with open(output_txt, "w", encoding="utf-8") as f:
    f.write("HOUSE ID DENGAN ATAP BAMBU\n")
    f.write("=" * 50 + "\n")
    f.write(f"Total rumah: {len(rumah_atap_bambu)}\n\n")

    if len(rumah_atap_bambu) > 0:
        f.write("Daftar house_id:\n")
        
        for house_id in rumah_atap_bambu["house_id"]:
            print(house_id)       # tampilkan di notebook
            f.write(f"{house_id}\n")  # simpan ke txt
    else:
        print("Tidak ada rumah dengan atap bambu.")
        f.write("Tidak ada rumah dengan atap bambu.\n")

print(f"\nOutput berhasil disimpan ke: {output_txt}")

# Tampilkan tabel di notebook
display(rumah_atap_bambu[["house_id", "house_type", "atap"]])
```
